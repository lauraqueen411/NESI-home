```python
import numpy as np
import pandas as pd
import xarray as xr
import matplotlib.pyplot as plt
import geopandas as gpd
import rioxarray
from scipy.stats import percentileofscore
import scipy
```

```python
# load streamflow data
flow_df = pd.read_csv('/nesi/project/niwa00015/queenle/data/flows/seasonal/40_50yrs/40plus.csv')
flow_df['date'] = pd.to_datetime(flow_df['date'])
flow_df = flow_df.set_index('date')
flow_df = flow_df.drop(['4901','1443499'],axis=1)
flow_df = flow_df['1979-03-01':'2019-03-01']

metadata_df = pd.read_csv('/nesi/project/niwa00015/queenle/data/metadata/40_50yr_sites.csv')
metadata_df.set_index('Number',inplace=True)
```

```python
# load catchment shape data
fp = '/nesi/project/niwa00015/queenle/data/arcgis/'
file = 'NESTCAT_v2_NZTM.shp'
catchment_data = gpd.read_file(fp+file)
catchment_data = catchment_data.to_crs("EPSG:4326")
catchment_data = catchment_data.loc[catchment_data['SITE'].isin(flow_df.columns)]
catchment_data.index = catchment_data['SITE']
```

```python
'''
Save csv files of all- and nat- ensemble simulations clipped to catchment resolution
'''

seasons_a = ['autumn','winter','spring','summer']

for scen in ['all','nat']:

    if scen == 'all':
        path = '~/../stoneda/data-for-laura/C20C/NOAA-ESRLandCIRES/ECHAM5-4/All-Hist/est1/v1-0/mon/land-ANZ/mrro/ens'
        file_name_setup = ['mrro_Lmon-ANZ_ECHAM5-4_All-Hist_est1_v1-0_ens','_197901-201902.nc']
    if scen == 'nat':
        path = '~/../stoneda/data-for-laura/C20C/NOAA-ESRLandCIRES/ECHAM5-4/Nat-Hist/obs-trend-1880s-est1/v1-0/mon/land-ANZ/mrro/ens'
        file_name_setup = ['mrro_Lmon-ANZ_ECHAM5-4_Nat-Hist_obs-trend-1880s-est1_v1-0_ens','_197901-201912.nc']

    for ens_mem in range(1,51):
        number = '0' + str(ens_mem) if ens_mem < 10 else str(ens_mem)
        file_path = path + number + "/" + file_name_setup[0]+number+file_name_setup[1]
        ens_mem = xr.open_dataset(file_path)

        # grab time window, resample to seasonal
        ens_mem = ens_mem['mrro'].sel(time=slice('1979-03-01','2019-03-01')).resample(time='QS-DEC').mean()

        # interpolate to finer resolution
        new_lon = np.linspace(ens_mem.lon[0], ens_mem.lon[-1], len(ens_mem.lon.values) * 30)
        new_lat = np.linspace(ens_mem.lat[0], ens_mem.lat[-1], len(ens_mem.lat.values) * 30)

        ens_mem_interp = ens_mem.interp(lat=new_lat, lon=new_lon,method='linear')

        for i,s in enumerate([3,6,9,12]):
            # select season
            season_ens_mem = ens_mem_interp.sel(time=ens_mem_interp.time.dt.month.isin([s]))

            season_ens_mem = season_ens_mem.rio.write_crs("EPSG:4326")
            season_ens_mem.rio.set_spatial_dims('lon', 'lat', inplace=True)

            site_df_list = []
            for site in flow_df:
                # clip by catchment
                catchment_geo = catchment_data.loc[[float(site)]]
                clipped = season_ens_mem.rio.clip(catchment_geo.geometry,catchment_data.crs, all_touched=True)

                # average over catchment
                site_df = (clipped.mean(['lon','lat'])).drop_vars('spatial_ref').to_dataframe().squeeze().rename(str(site))

                site_df_list.append(site_df)

            ens_mem_catchments = pd.concat(site_df_list,axis=1)

            #ens_mem_catchments.to_csv('/nesi/project/niwa00015/queenle/data/hydro_fingerprinting/model/raw/ECHAM5_' + scen + "_" + seasons_a[i] + "_ens" + number + ".csv")


```

```python
'''
Write csv files for each ensemble member transformed to gaussian space using quantile mapping based on 
full ensemble distribution
'''

seasons_a = ['autumn','winter','spring','summer']

    
for scen in ['all','nat']:
    
    for i,s in enumerate([3,6,9,12]):

        # create concatenated (full ensemble) all-forcings simulations for full distribution
        ens_df_list = []
        for ens_mem in range(1,51):
            number = '0' + str(ens_mem) if ens_mem < 10 else str(ens_mem)
            ens_mem = pd.read_csv('/nesi/project/niwa00015/queenle/data/hydro_fingerprinting/model/raw/ECHAM5_'+scen+'_'+seasons_a[i]+'_ens'+number+'.csv')
            ens_mem.time = pd.to_datetime(ens_mem.time)
            ens_mem.set_index('time', inplace=True)
            #ens_mem = ens_mem - ens_mem.mean()

            ens_df_list.append(ens_mem)

        concatenated_ens = pd.concat(ens_df_list)

        # map each ensemble member to gaussian space using full ensemble all-forcing simulations
        for ens_mem in range(1,51):
            number = '0' + str(ens_mem) if ens_mem < 10 else str(ens_mem)
            ens_mem = pd.read_csv('/nesi/project/niwa00015/queenle/data/hydro_fingerprinting/model/raw/ECHAM5_'+scen+'_'+seasons_a[i]+'_ens'+number+'.csv')
            ens_mem.time = pd.to_datetime(ens_mem.time)
            ens_mem.set_index('time', inplace=True)
            #ens_mem = ens_mem - ens_mem.mean()

            gaussian_df = ens_mem.copy()
            for site in ens_mem:

                all_dist = concatenated_ens[site]
                member = ens_mem[site]

                corrected = []
                for j, val in enumerate(member):
                    percentile = percentileofscore(all_dist, val, kind='mean')
                    corrected.append(scipy.stats.norm.ppf(percentile/100, loc=0, scale=1))

                gaussian_df[site] = corrected

            gaussian_df.to_csv('/nesi/project/niwa00015/queenle/data/hydro_fingerprinting/model/gaussian/ECHAM5_'+scen+'_'+seasons_a[i]+'_ens'+number+'.csv')

```

```python
'''
Create ensemble mean csv files for all- and nat-forcing scenarios
'''
seasons_a = ['autumn','winter','spring','summer']

for scen in ['all','nat']:
    for i,s in enumerate([3,6,9,12]):  
        ens_df_list = []
        for ens_mem in range(1,51):
            number = '0' + str(ens_mem) if ens_mem < 10 else str(ens_mem)
            ens_mem = pd.read_csv('/nesi/project/niwa00015/queenle/data/hydro_fingerprinting/model/gaussian/ECHAM5_'+scen+'_'+seasons_a[i]+'_ens'+number+'.csv')
            ens_mem.time = pd.to_datetime(ens_mem.time)
            ens_mem.set_index('time', inplace=True)

            ens_df_list.append(ens_mem)

        ens_mean = sum(ens_df_list)/len(ens_df_list)
        ens_mean.to_csv('/nesi/project/niwa00015/queenle/data/hydro_fingerprinting/model/gaussian/ECHAM5_'+scen+'_'+seasons_a[i]+'_ensMean.csv')

    
```
