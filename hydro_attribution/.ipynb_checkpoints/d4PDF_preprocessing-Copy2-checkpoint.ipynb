```python
import numpy as np
import pandas as pd
import xarray as xr
from functools import reduce
import matplotlib.pyplot as plt
from idlpy import *
from scipy import stats
import statsmodels.api as sm
from scipy.stats import percentileofscore
import scipy

import seaborn as sns
import cartopy.crs as ccrs
```

```python
def q_map_to_gaussian(da, dist):
    
    # create container for corrected values
    gaussian = []
    
    for lat in da.lat:
        for lon in da.lon:

            ts = da.sel(lat=lat,lon=lon).values
            full_dist = dist.sel(lat=lat,lon=lon).values
            
            if np.isnan(np.sum(ts)):
                corrected_ts = [np.nan for i in range(len(ts))]

            else:
                corrected_ts = []
                for val in ts:
                    
                    # find percentile of value within the distribution of observations
                    percentile = percentileofscore(full_dist, val, kind= 'mean')
                    
                    if percentile == 100 or percentile == 0:
                        print('issue')

                    # replace with corresponding percentile value in norm distribution
                    corrected_ts.append(scipy.stats.norm.ppf(percentile/100, loc=0, scale=1))

            gaussian.append(corrected_ts)

    gaussian = np.array(gaussian).reshape((len(da.lat),len(da.lon),len(da.time)))
    
    gaussian_da = xr.DataArray(data = gaussian,
                               dims = ('lat', 'lon','time'),
                               coords = {'lat':da.lat.values,
                                         'lon':da.lon.values,
                                         'time':da.time.values
                                        }
                               )
    return(gaussian_da)
```

```python
'''
d4PDF land-sea mask
'''
example = xr.open_dataset('/nesi/project/niwa02986/deansm/data/d4pdf/NZ/HPB/mon/TA/TA_Aday_d4PDF_HPB_m001_1951-2011.nc')
# d4PDF land-sea mask
lsm = xr.open_dataset('/nesi/project/niwa02986/queenle/data/d4pdf/sftlf_fx_d4PDF_HPB_m000_000000-000000.nc')
lsm = lsm.isel(time=0).drop('time')
lsm = lsm.sel(lon = slice(165,180),lat=slice(-48,-33))
lsm = lsm.interp(lat=example.lat,lon=example.lon)
```

```python
'''
Creating dataarrays of FULL distributions for ALL and NAT scenarios (concatenate all ensemble members over time)
'''

lower_var = {'TA':'ta','PRECIPI':'precipi','ROFS':'rofs'}

for var in ['TA','PRECIPI']:#,'ROFS']:
    
    # ALL SCENARIO
    path = '/nesi/project/niwa02986/deansm/data/d4pdf/NZ/HPB/mon/'+var+'/'
    file_name_setup = [var+'_Aday_d4PDF_HPB_m','_1951-2011.nc']

    seasonal_das = {3:[],6:[],9:[],12:[]}
    for ens_mem in range(1,101):
        if ens_mem == 100:
            number = '100'
        else:
            number = '00' + str(ens_mem) if ens_mem < 10 else '0' + str(ens_mem)

        file_path = path + file_name_setup[0]+number+file_name_setup[1]
        ens_mem = xr.open_dataset(file_path)

        ens_mem = ens_mem.sel(lon = slice(165,180),lat=slice(-48,-33))
        ens_mem = ens_mem[lower_var[var]].where(lsm.sftlf > 0)

        ens_mem_seasonal = ens_mem.resample(time='QS-DEC').mean()

        for s in [3,6,9,12]:
            ens_mem_season = ens_mem_seasonal.sel(time=ens_mem_seasonal.time.dt.month.isin([s]))

            seasonal_das[s].append(ens_mem_season)

    for i,s in enumerate([3,6,9,12]):
        season_full_dist = xr.concat(seasonal_das[s],dim="time")
        season_full_dist.to_netcdf("/nesi/project/niwa00015/queenle/data/hydro_fingerprinting/model/d4PDF/full_dists/"+var+"/" + var + "_all_" + seasons_a[i] + "_full_distribution.nc")
        
    # NAT SCENARIO
    path = '/nesi/project/niwa02986/deansm/data/d4pdf/NZ/HPB_NAT/mon/'+var+'/'
    file_name_setup = [var+'_Aday_d4PDF_HPB_NAT_m','_1951-2010.nc']

    seasonal_das = {3:[],6:[],9:[],12:[]}
    for ens_mem in range(1,101):
        if ens_mem == 100:
            number = '100'
        else:
            number = '00' + str(ens_mem) if ens_mem < 10 else '0' + str(ens_mem)

        file_path = path + file_name_setup[0]+number+file_name_setup[1]
        ens_mem = xr.open_dataset(file_path)

        ens_mem = ens_mem.sel(lon = slice(165,180),lat=slice(-48,-33))
        ens_mem = ens_mem[lower_var[var]].where(lsm.sftlf > 0)

        ens_mem_seasonal = ens_mem.resample(time='QS-DEC').mean()

        for s in [3,6,9,12]:
            ens_mem_season = ens_mem_seasonal.sel(time=ens_mem_seasonal.time.dt.month.isin([s]))

            seasonal_das[s].append(ens_mem_season)

    for i,s in enumerate([3,6,9,12]):
        season_full_dist = xr.concat(seasonal_das[s],dim="time")
        season_full_dist.to_netcdf("/nesi/project/niwa00015/queenle/data/hydro_fingerprinting/model/d4PDF/full_dists/"+var+"/" + var + "_nat_" + seasons_a[i] + "_full_distribution.nc")
        
```

```python
'''
gaussian-ifying annual and seasonal pr for each ensemble member
'''

seasons_a = ['autumn','winter','spring','summer']
lower_var = {'TA':'ta','PRECIPI':'precipi','ROFS':'rofs'}

for var in ['PRECIPI']:#,'PRECIPI','ROFS']:
    print(var)
    for scen in ['all','nat']:
        print(scen)
        
        if scen == 'all':
            path = '/nesi/project/niwa02986/deansm/data/d4pdf/NZ/HPB/mon/'+var+'/'
            file_name_setup = [var+'_Aday_d4PDF_HPB_m','_1951-2011.nc']
        if scen == 'nat':
            path = '/nesi/project/niwa02986/deansm/data/d4pdf/NZ/HPB_NAT/mon/'+var+'/'
            file_name_setup = [var+'_Aday_d4PDF_HPB_NAT_m','_1951-2010.nc']

        for ens_mem in range(1,101):

            print(ens_mem)

            if ens_mem == 100:
                number = '100'
            else:
                number = '00' + str(ens_mem) if ens_mem < 10 else '0' + str(ens_mem)

            file_path = path + file_name_setup[0]+number+file_name_setup[1]
            ens_mem = xr.open_dataset(file_path)

            ens_mem = ens_mem.sel(lon = slice(165,180),lat=slice(-48,-33))
            ens_mem = ens_mem[lower_var[var]].where(lsm.sftlf > 0)

            ens_mem_seasonal = ens_mem.resample(time='QS-DEC').mean()

            for i,s in enumerate([3,6,9,12]):

                full_seasonal_dist = xr.open_dataset("/nesi/project/niwa00015/queenle/data/hydro_fingerprinting/model/d4PDF/full_dists/"+var+"/" + var + "_"+scen+"_" + seasons_a[i] + "_full_distribution.nc")
                full_seasonal_dist = full_seasonal_dist[lower_var[var]]
                #print(full_seasonal_dist)
                
                ens_mem_season = ens_mem_seasonal.sel(time=ens_mem_seasonal.time.dt.month.isin([s]))

                ens_mem_season_gaussian = q_map_to_gaussian(ens_mem_season,full_seasonal_dist)
                ens_mem_season_gaussian.to_netcdf("/nesi/project/niwa00015/queenle/data/hydro_fingerprinting/model/d4PDF/gaussian/"+var+"/" + file_name_setup[0] + number + "_gaussian_" + seasons_a[i] + file_name_setup[1])
                ens_mem_season.to_netcdf("/nesi/project/niwa00015/queenle/data/hydro_fingerprinting/model/d4PDF/raw/"+var+"/" + file_name_setup[0] + number + "_raw_" +seasons_a[i] + file_name_setup[1])
                
```

```python
'''
Create full ensemble means
'''

for var in ['TA','PRECIPI']:#'ROFS'
    for scen in ['all','nat']:

        if scen == 'all':
            file_name_setup = [var+'_Aday_d4PDF_HPB_m','_1951-2011.nc']
        if scen == 'nat':
            file_name_setup = [var+'_Aday_d4PDF_HPB_NAT_m','_1951-2010.nc']
        
        for i,s in enumerate([3,6,9,12]):

            da_list = []
            for ens_mem in range(1,101):
                if ens_mem == 100:
                    number = '100'
                else:
                    number = '00' + str(ens_mem) if ens_mem < 10 else '0' + str(ens_mem)
                if number == '099':
                    continue
                    
                ens_mem_season = xr.open_dataarray("/nesi/project/niwa00015/queenle/data/hydro_fingerprinting/model/d4PDF/gaussian/"+var+"/" + file_name_setup[0] + number + "_gaussian_" + seasons_a[i] + file_name_setup[1])

                da_list.append(ens_mem_season)

            season_ens_mean = sum(da_list)/len(da_list)
            season_ens_mean.to_netcdf('/nesi/project/niwa00015/queenle/data/hydro_fingerprinting/model/d4PDF/gaussian/'+var+'/means/' + file_name_setup[0][:-1] + 'FullEnsMean' + "_gaussian_" + seasons_a[i] + file_name_setup[1])

            for ens_mem in range(1,101):
                if ens_mem == 100:
                    number = '100'
                else:
                    number = '00' + str(ens_mem) if ens_mem < 10 else '0' + str(ens_mem)
                if number == '099':
                    continue
                    
                temp = da_list.copy()
                temp.pop(ens_mem-1)
                
                mean_sub_mem = sum(temp)/len(temp)
                mean_sub_mem.to_netcdf('/nesi/project/niwa00015/queenle/data/hydro_fingerprinting/model/d4PDF/gaussian/'+var+'/means/' + file_name_setup[0][:-1] + 'EnsMeanSub_m' + number + "_gaussian_" + seasons_a[i] + file_name_setup[1])
```
