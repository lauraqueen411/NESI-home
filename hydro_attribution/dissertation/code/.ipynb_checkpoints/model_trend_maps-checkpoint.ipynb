```python
import xarray as xr
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
import cartopy.crs as ccrs
import cartopy.feature as cf
```

```python
var_names = ['temp','precip','runoff']
var_dict = {'ECHAM5':['tas','pr','mrro'],
            'HadGEM3':['tas','pr'],
            'd4PDF':['TA','PRECIPI','ROF']}

time_window_dict = {'ECHAM5':('1979','2018'),
                    'HadGEM3':('1961','2015'),
                    'd4PDF':('1951','2021')}

v_info = [0.035,0.01,0.01]
units = ['C/']

cb_config = [[-0.035,-0.0175,0,0.0175,0.035],[-0.01,-0.005,0,0.005,0.01],[-0.01,-0.005,0,0.005,0.01]]

font = {'family': 'sans-serif',
        'color':  'black',
        'weight': 'normal',
        'size': 22,
        }

season_titles = {'winter':'Winter','spring':'Spring','summer':'Summer','autumn':'Autumn'}

input_dir = '/nesi/project/niwa00015/queenle/data/hydro_fingerprinting/model/'
```

```python
def get_linregress_slope(arr):
    if (np.isnan(arr)).any():
        result = np.nan
    else:
        result = np.polyfit([i for i in range(len(arr))],arr,deg=1)[0]
        
    return(result)
```

```python
model = 'HadGEM3'
var = 'tas'
da = xr.open_dataarray(input_dir+model+'/gaussian/'+var+'/seasonal/means/'+model+'_all_'+season+'_'+var+'_gaussian_'+'ensMean'+'.nc')
```

```python
da.mean('time').plot()
```

```python

season = 'winter'
for var_index in [0,1,2]:
    
    
    if var_index < 2:
        fig,axs = plt.subplots(2,3,figsize=(10,6),sharex=True,sharey=True,subplot_kw=dict(projection=ccrs.PlateCarree()),layout='constrained')
        models = ['HadGEM3','ECHAM5','d4PDF']

    else:
        fig,axs = plt.subplots(2,2,figsize=(7,6),sharex=True,sharey=True,subplot_kw=dict(projection=ccrs.PlateCarree()),layout='constrained')
        models = ['ECHAM5','d4PDF']

    for j,model in enumerate(models):

        try:
            var = var_dict[model][var_index]
        except:
            continue

        all_da = xr.open_dataarray(input_dir+model+'/gaussian/'+var+'/seasonal/means/'+model+'_all_'+season+'_'+var+'_gaussian_'+'ensMean'+'.nc')#.sel(time=slice('1979','2015'))
        nat_da = xr.open_dataarray(input_dir+model+'/gaussian/'+var+'/seasonal/means/'+model+'_nat_'+season+'_'+var+'_gaussian_'+'ensMean'+'.nc')#.sel(time=slice('1979','2015'))
        ant_da = all_da - nat_da

        for i,da in enumerate([ant_da,nat_da]):
            im = xr.apply_ufunc(get_linregress_slope, da, input_core_dims=[["time"]], vectorize = True).plot(ax=axs[i][j],transform=ccrs.PlateCarree(),vmin=-v_info[var_index],vmax=v_info[var_index],cmap='RdBu_r',add_colorbar=False)

            axs[i][j].coastlines()
            axs[i][j].set_extent([164,180, -48,-33])#, crs=noproj)

            if i == 0:
                axs[i][j].set_title(model + '\n(' + time_window_dict[model][0] + '-' + time_window_dict[model][1] + ')',fontsize=18)

            if j == 0:
                axs[i][j].text(-0.05,0.5,['ANT','NAT'][i],transform=axs[i][j].transAxes,rotation=90, fontsize=18, va='center', ha='center')


    cb = fig.colorbar(im,ax=axs[:,:],shrink=0.85)
    cb.set_label(f'Normalized {var_names[var_index]} trend (/yr)',fontsize=18)
    cb.set_ticks(cb_config[var_index],labels=[str(val) for val in cb_config[var_index]],fontsize=14)
    
    plt.savefig('/home/queenle/hydro_attribution/dissertation/figures/supplemental_material/signal_maps_'+var_names[var_index]+'.png',dpi=300)
    plt.close()
```
