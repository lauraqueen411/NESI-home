```python
FWI = d4pdf_FWI.isel(latitude=8,longitude=33).FWI.resample(time='A').mean()
plt.scatter([i for i in range(len(FWI))],FWI,s=5)

d4pdf_poly_fit = FWI.polyfit(dim='time',deg=1,skipna=True)

intercept = d4pdf_poly_fit.polyfit_coefficients.sel(degree=0).values[0]#* 1e9 * 60 * 60 * 24
slope = (d4pdf_poly_fit.polyfit_coefficients.sel(degree=1).values[0]) * 1e9 * 60 * 60 * 24

reg = [slope*i + intercept for i in np.linspace(0,60,num=len(FWI))]
plt.plot(reg,color='red')
```

```python
'''
MEAN comparison plots
'''

names = {'t2m':['t2m','tasmax'],
        'rh':['LN rh','daily min rh'],
        'wind':['LN wind','daily sfwind'],
        'tp':['tp','tp']}

ranges = {'t2m':[3,33],
        'rh':[0,100],
        'wind':[6,26],
        'tp':[0,10]}

for var in ['tp','t2m','rh','wind']:

    fig,axs = plt.subplots(2,2,figsize=(14,12))
    axs = axs.flatten()

    ERA5_var = ERA5_FWI_inputs[var]
    d4pdf_var = d4pdf_FWI_inputs[var]
    
    # MEAN
    ERA5_var.mean('time').plot(ax=axs[0],vmin=ranges[var][0],vmax=ranges[var][1])
    d4pdf_var.mean('time').plot(ax=axs[1],vmin=ranges[var][0],vmax=ranges[var][1])

    axs[0].set_title("ERA5 mean "+names[var][0]+" (1979-2022)")
    axs[1].set_title("d4pdf mean "+names[var][1]+" (1951-2011)")

    
    #TREND
    shared_ERA5 = ERA5_var.sel(time=slice('1979','2011'))
    shared_d4pdf = d4pdf_var.sel(time=slice('1979','2011'))

    ERA5_poly_fit = shared_ERA5.polyfit(dim='time',deg=1,skipna=True)
    ERA5_poly_fit.sel(degree=1).polyfit_coefficients.plot(ax=axs[2])#,vmin=-1.5e-17,vmax=1.5e-17,cmap='RdBu_r')

    d4pdf_poly_fit = shared_d4pdf.polyfit(dim='time',deg=1,skipna=True)
    d4pdf_poly_fit.sel(degree=1).polyfit_coefficients.plot(ax=axs[3])#,vmin=-1.5e-17,vmax=1.5e-17,cmap='RdBu_r')

    axs[2].set_title("ERA5 "+names[var][0]+" trend (1979-2011)")
    axs[3].set_title("d4pdf "+names[var][1]+" trend (1979-2011)")

    #plt.savefig("/nesi/project/niwa00015/queenle/plots/fire/"+var+"_ERA5_d4df_compare.png",dpi=300)
```

```python
'''
FWI comparison plots
'''

fig,axs = plt.subplots(2,2,figsize=(14,12))
axs = axs.flatten()

ERA5_var = ERA5_FWI.FWI
d4pdf_var = d4pdf_FWI.FWI

# MEAN
ERA5_var.mean('time').plot(ax=axs[0])
d4pdf_var.mean('time').plot(ax=axs[1])

axs[0].set_title("ERA5 mean FWI (1979-2022)")
axs[1].set_title("d4pdf mean FWI (1951-2011)")


#TREND
shared_ERA5 = ERA5_var.sel(time=slice('1979','2011'))
shared_d4pdf = d4pdf_var.sel(time=slice('1979','2011'))

ERA5_poly_fit = shared_ERA5.polyfit(dim='time',deg=1,skipna=True)
ERA5_poly_fit.sel(degree=1).polyfit_coefficients.plot(ax=axs[2])#,vmin=-1.5e-17,vmax=1.5e-17,cmap='RdBu_r')

d4pdf_poly_fit = shared_d4pdf.polyfit(dim='time',deg=1,skipna=True)
d4pdf_poly_fit.sel(degree=1).polyfit_coefficients.plot(ax=axs[3])#,vmin=-1.5e-17,vmax=1.5e-17,cmap='RdBu_r')

axs[2].set_title("ERA5 FWI trend (1979-2011)")
axs[3].set_title("d4pdf FWI trend (1979-2011)")

plt.savefig("/nesi/project/niwa00015/queenle/plots/fire/FWI_ERA5_d4df_compare.png",dpi=300)
```
