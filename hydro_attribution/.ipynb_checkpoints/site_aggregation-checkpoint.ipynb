Site Aggregation

This workbook overlays the observed flow sites atop the model grid, converting the observed data to the model's spatial dimension. 

New seasonal streamflow datasets for the obs and ECHAM data are created containing runoff at shared cells. These datasets are written to files as ~/data/runoff/ as season_scen (i.e. fall_obs.csv, fall_All.csv, etc.).

```python
# read in example ECHAM flow data
file_name = 'mrro_Lmon-ANZ_ECHAM5-4_All-Hist_est1_v1-0_ens01_197901-201902.nc'
file_path = '../../stoneda/data-for-laura/C20C/NOAA-ESRLandCIRES/ECHAM5-4/All-Hist/est1/v1-0/mon/land-ANZ/mrro/ens01/' + file_name

example_echam = xr.open_dataset(file_path)['mrro']

# read in site metadata
metadata_df = pd.read_csv('/nesi/project/niwa00015/queenle/data/metadata/40_50yr_sites.csv')
metadata_df.set_index('Number',inplace=True)
```

```python
# convert obs flow (L/s) to kg/m2/s by dividing by catchment area

flow_df = pd.read_csv('/nesi/project/niwa00015/queenle/data/flows/seasonal/40_50yrs/40plus.csv')

flow_df.date = pd.to_datetime(flow_df.date)
flow_df.set_index('date', inplace=True)

flow_df = flow_df.loc['1979':'2019']
#flow_df = flow_df.mask(flow_df==0).fillna(flow_df.mean())

'''
for col in flow_df.columns:
    area = metadata_df.loc[int(col)]['area_km2']
    area_m2 = area*1000000
    flow_df[col] = flow_df[col] / area_m2
'''
```

```python
'''
record which cell an obs_site belongs in the following containers:
site_cellnums: a list of int to be used as color information in the following plot
aggregation_dict: a dict {(cell coords):[list of site numbers]} which records cells 
and the sites they contain.

the site allocation is determined by distance: the cell coord closest to the site's 
centroid coord is that site's cell
'''

model_data = example_echam.mean('time')

mo_lats = model_data.lat.values
mo_lons = model_data.lon.values

site_cellnums = []
aggregation_dict = {}
for site in metadata_df.index:

    ob_lat = metadata_df.loc[site]['centroid_lat']
    ob_lon = metadata_df.loc[site]['centroid_lon']
    
    mn = 100000
    cell_num = 0
    count = 0
    for la in mo_lats:
        for lo in mo_lons:
            if model_data.sel(lat=la,lon=lo) == 0:
                #print('here')
                continue

            coord = str(la) + " " + str(lo)
            if coord not in aggregation_dict:
                aggregation_dict[coord] = []
                
            dist = distance((ob_lat,ob_lon),(la,lo))
            if dist < mn:
                mn = dist
                cell_coords = coord
                cell_num = count
            count += 1
    
    site_cellnums.append(cell_num)
    aggregation_dict[cell_coords].append(site)
      
```

```python
# plot the obs sites atop the model grid and color by cell 'ownership'.

plt.figure(figsize=(22,18))
ax=plt.axes(projection=ccrs.PlateCarree())

example_echam.mean('time').plot.pcolormesh(ax=ax, transform=ccrs.PlateCarree(),add_colorbar=False,cmap='bwr',vmin=-0.3940715e-06,vmax=0.3940715e-06,alpha=.2)
ax.scatter([val for val in metadata_df['centroid_lon'].values],[val for val in metadata_df['centroid_lat'].values],c=[colors[val+5] for val in site_cellnums])

ax.coastlines(resolution='50m', color='black', linewidth=1)
ax.set_xlim([164,180])

plt.show()
```

```python
'''
Create new seasonal datasets of observed flow converted to cell-space.
These datasets are written to files in ~/data/runoff/ as season_obs.csv.
'''

seasons = [3,6,9,12]
season_dfs = []
done = False

for season in seasons:
    
    # list of cell dataframes which contain the average seasonal flows across the sites located in the cell
    cell_dfs = []
    
    for cell in aggregation_dict:

        # continue if cell does not contain any obs sites
        if aggregation_dict[cell] == []:
            continue
            
        # df including only sites in cell
        sites_to_merge = flow_df[[str(site) for site in aggregation_dict[cell]]]
                
        # filter out season
        sites_by_season = sites_to_merge[sites_to_merge.index.month.isin([season])]
                
        # df of average flow of all sites in cell
        cell_sum = sites_by_season.sum(axis=1,skipna=True).to_frame()
                
        joined_area = 0
        for col in sites_to_merge.columns:
            area = metadata_df.loc[int(col)]['area_km2']
            area_m2 = area*1000000
            joined_area += area_m2
            
        cell_average = cell_sum / joined_area
        
        # name average cell coordinate
        cell_average.columns = [cell]
                
        cell_dfs.append(cell_average)
    
    # merge all cell average dfs into one season observations df
    season_obs = reduce(lambda  left,right: pd.merge(left,right,on=['date'],how='outer'), cell_dfs)
   
    season_dfs.append(season_obs)
    
fall_obs,winter_obs,spring_obs,summer_obs = season_dfs

fall_obs.to_csv('/nesi/project/niwa00015/queenle/data/runoff/fall_obs.csv')
winter_obs.to_csv('/nesi/project/niwa00015/queenle/data/runoff/winter_obs.csv')
spring_obs.to_csv('/nesi/project/niwa00015/queenle/data/runoff/spring_obs.csv')
summer_obs.to_csv('/nesi/project/niwa00015/queenle/data/runoff/summer_obs.csv')
```
