```python
import xarray as xr
from datetime import datetime
import math
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from scipy import stats
```

```python
hr_pr_FWI = xr.open_dataset("/nesi/project/niwa00015/queenle/data/fire/d4pdf/testing_hr_v_day_pr/d4pdf_FWI_hrly_pr.nc")
day_pr_FWI = xr.open_dataset("/nesi/project/niwa00015/queenle/data/fire/d4pdf/testing_hr_v_day_pr/d4pdf_FWI_daily_pr.nc")
```

```python
hr_pr_FWI.FWI.mean('time').plot()
```

```python
day_pr_FWI.FWI.mean('time').plot()
```

```python
m = hr_pr_FWI.FWI.sel(latitude=40,longitude=-120,method='nearest').resample(time='M').mean()
m.sel(time = m.time.dt.month.isin([7])).plot(c='red')

m = day_pr_FWI.FWI.sel(latitude=40,longitude=-120,method='nearest').resample(time='M').mean()
m.sel(time = m.time.dt.month.isin([7])).plot(c='blue')
```
