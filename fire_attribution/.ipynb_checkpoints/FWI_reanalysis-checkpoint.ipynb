```python
'''
Pearson's r
'''

import scipy

def calc_pearsons_r(da):
    
    lats = []
    lons = []
    rs = []
    ps = []

    for lat in da.Latitude.values:
        for lon in da.Longitude.values:
            cell_da = da.sel(Latitude=lat,Longitude=lon)

            x = cell_da.Time.dt.year.values.tolist()
            y = cell_da.values.tolist()

            if np.isnan(y).any():
                r,p = [math.nan,math.nan]

            else:
                r,p = scipy.stats.pearsonr(x,y)

            lats.append(lat)
            lons.append(lon)
            rs.append(r)
            ps.append(p)

    r_df = pd.DataFrame({"rs":rs,"lat":lats, "lon":lons})
    p_df = pd.DataFrame({"ps":ps,"lat":lats, "lon":lons})
    
    idx = pd.MultiIndex.from_arrays(arrays=[lats,lons], names=["lat","lon"])
    
    r_s = pd.Series(data=rs, index=idx)
    p_s = pd.Series(data=ps, index=idx)
    
    rs_da = xr.DataArray.from_series(r_s)
    rs_da = rs_da.rename('r')

    ps_da = xr.DataArray.from_series(p_s)
    ps_da = ps_da.rename('p')
    
    return([rs_da,ps_da])
```

```python
import cartopy.crs as ccrs

for var in ['FWI','BUI','DC','DMC','FFMC','DSR','ISI']:
    
    fig, ax = plt.subplots(figsize=(12,8), subplot_kw={'projection': ccrs.PlateCarree()})
    
    [r,p] = calc_pearsons_r(July_components[var])
    
    r.plot(ax=ax)
    ax.set_title(var)
    
    plt.show()
```

```python
'''
linear regression slope
'''

lats = []
lons = []
slopes = []

for lat in July_FWI.Latitude.values:
    for lon in July_FWI.Longitude.values:
        cell_da = July_FWI.sel(Latitude=lat,Longitude=lon)

        x = cell_da.Time.dt.year.values.tolist()
        y = cell_da.values.tolist()

        m,b = np.polyfit(x, y, 1)
        
        lats.append(lat)
        lons.append(lon)
        slopes.append(m)
        
df = pd.DataFrame({"slopes":slopes, "lat":lats, "lon":lons})

idx = pd.MultiIndex.from_arrays(arrays=[lats,lons], names=["lat","lon"])
s = pd.Series(data=slopes, index=idx)

linreg_da = xr.DataArray.from_series(s)
linreg_da = linreg_da.rename('m')
```
