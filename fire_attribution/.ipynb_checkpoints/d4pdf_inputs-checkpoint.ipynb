```python
tasmax_ds['lon'] = np.array([lon-360 for lon in tasmax_ds.lon])
rhmin_ds['lon'] = np.array([lon-360 for lon in rhmin_ds.lon])
wind_ds['lon'] = np.array([lon-360 for lon in wind_ds.lon])
pr_hr_ds['lon'] = np.array([lon-360 for lon in pr_hr_ds.lon])
pr_day_ds['lon'] = np.array([lon-360 for lon in pr_day_ds.lon])
lsm['lon'] = np.array([lon-360 for lon in lsm.lon])
```

```python
'''
CREATE LAND/TIME ZONE MASKS
'''

tz1 = tz_data.loc[tz_data.tz_name1st=='America/Los_Angeles'].geometry #zone=-8.0; tz_name1st = America/Los_Angeles
tz1_noon_utc = 12 - tz_data.loc[tz_data.tz_name1st=='America/Los_Angeles'].zone.values[0] 
tz1_mask = lsm.rio.clip(tz1, tz1.crs,drop=False).fillna(0)
tz1_mask = tz1_mask.sel(lat=slice(tasmax_ds.lat[0],tasmax_ds.lat[-1]),lon=slice(tasmax_ds.lon[0],tasmax_ds.lon[-1]))
tz1_mask = tz1_mask.interp(lat=tasmax_ds.lat,lon=tasmax_ds.lon)

tz2 = tz_data.loc[tz_data.tz_name1st=='America/Denver'].geometry 
tz2_noon_utc = 12 - tz_data.loc[tz_data.tz_name1st=='America/Denver'].zone.values[0]
tz2_mask = lsm.rio.clip(tz2, tz2.crs,drop=False).fillna(0)
tz2_mask = tz2_mask.sel(lat=slice(tasmax_ds.lat[0],tasmax_ds.lat[-1]),lon=slice(tasmax_ds.lon[0],tasmax_ds.lon[-1]))
tz2_mask = tz2_mask.interp(lat=tasmax_ds.lat,lon=tasmax_ds.lon)
```

```python
# calculate 24hr accumulations from local noon for both time zones

# convert kg/m2/s to mm/hr, sum over 24 hours
pr_hr_da = pr_hr_ds.precipi*(60*60)

tz1_tp = pr_hr_da.resample(time="24H", base = tz1_noon_utc, closed="right",label="right").sum().where(tz1_mask.sftlf>0, other=0)
tz1_tp = tz1_tp.resample(time='1D').first()

tz2_tp = pr_hr_da.resample(time="24H", base = tz2_noon_utc, closed="right",label="right").sum().where(tz2_mask.sftlf>0, other=0)
tz2_tp = tz2_tp.resample(time='1D').first()

tp_da = tz1_tp + tz2_tp
```

```python
# remove time from other variables
tasmax_da = tasmax_ds.ta.resample(time='1D').first()
rhmin_da = rhmin_ds.rha.resample(time='1D').first()
sws_da = wind_ds.wind.resample(time='1D').first()

# merge variables into one dataset
FWI_inputs = xr.merge([tp_da, tasmax_da, rhmin_da, sws_da])
FWI_inputs = FWI_inputs.where((tz1_mask + tz2_mask).sftlf > 0)
FWI_inputs = FWI_inputs.rename({'ta':'t2m','rha':'rh','precipi':'tp'})
```

```python
'''
CONVERT UNITS
'''

# temp K to C
FWI_inputs["t2m"] = FWI_inputs["t2m"] - 273.15 # convert temp K to C

# wind m/s to km/h
FWI_inputs["wind"] = (FWI_inputs["wind"] * 60 * 60)/1000

```

```python


# merge variables into one dataset
new_FWI_inputs = xr.merge([pr_day_ds.precipi, tasmax_da, rhmin_da, sws_da])
new_FWI_inputs = new_FWI_inputs.where((tz1_mask + tz2_mask).sftlf > 0)
new_FWI_inputs = new_FWI_inputs.rename({'ta':'t2m','rha':'rh','precipi':'tp'})

'''
CONVERT UNITS
'''

# temp K to C
new_FWI_inputs["t2m"] = new_FWI_inputs["t2m"] - 273.15 # convert temp K to C

# wind m/s to km/h
new_FWI_inputs["wind"] = (new_FWI_inputs["wind"] * 60 * 60)/1000

new_FWI_inputs["tp"] = new_FWI_inputs["tp"]*(60*60*24)
```
