```python
import numpy as np
import pandas as pd
import xarray as xr
import matplotlib.pyplot as plt
```

```python
font = {'family': 'sans-serif',
        'color':  'black',
        'weight': 'normal',
        'size': 22,
        }

season_titles = {'winter':'Winter','spring':'Spring','summer':'Summer','autumn':'Autumn'}
```

```python
'''
load result files
'''

result_dir = '/nesi/project/niwa00015/queenle/data/hydro_fingerprinting/model/ECHAM5/summary_csv/'


tas_dfs = {}

# no processing results
no_proc_df = pd.read_csv('/nesi/project/niwa00015/queenle/results/ECHAM5_PM_results_final.csv')
no_proc_df = no_proc_df.query('variable==\''+'tas'+'\' and trunc==\''+str(36)+'\' and time_step==\''+'annual'+'\'')

detrend_nat_df = pd.read_csv('/nesi/project/niwa00015/queenle/results/ECHAM5_tas_perfect_model_results_detrended_NAT.csv')
detrend_nat_df = detrend_nat_df.query('variable==\''+'tas'+'\' and trunc==\''+str(36)+'\' and time_step==\''+'annual'+'\'')

detrend_all_and_nat_df = pd.read_csv('/nesi/project/niwa00015/queenle/results/ECHAM5_tas_perfect_model_results_detrended_ALL_detrended_NAT.csv')
detrend_all_and_nat_df = detrend_all_and_nat_df.query('variable==\''+'tas'+'\' and trunc==\''+str(36)+'\' and time_step==\''+'annual'+'\'')


tas_dfs['No processing NAT trends'] = no_proc_df
tas_dfs['NAT trend removed from ALL + NAT'] = detrend_all_and_nat_df
tas_dfs['NAT trend assigned to ANT'] = detrend_nat_df
```

```python
fig,axs = plt.subplots(4,3,figsize=(12,12),sharex=True,sharey=True)

for i,season in enumerate(['winter','spring','summer','autumn']):
    
    for j,scen in enumerate(tas_dfs.keys()):
        
        df = tas_dfs[scen].query('season==\''+season+'\'').sort_values('ant_best')

        ax = axs[i][j]

        x_vals = [i for i in range(len(df['ensemble_member']))]

        ax.scatter(x_vals,df['ant_best'],color='red',marker='.',label='ANT')
        ax.scatter(x_vals,df['ant_upper'],color='red',marker='_')
        ax.scatter(x_vals,df['ant_lower'],color='red',marker='_')
        ax.vlines(x_vals,df['ant_lower'],df['ant_upper'],color='red')

        ax.scatter(x_vals,df['nat_best'],color='blue',marker='.',label='NAT')
        ax.scatter(x_vals,df['nat_upper'],color='blue',marker='_')
        ax.scatter(x_vals,df['nat_lower'],color='blue',marker='_')
        ax.vlines(x_vals,df['nat_lower'],df['nat_upper'],color='blue')

        ax.axhline(0,-0.5,1.5,c='black',alpha=0.8)
        ax.axhline(1,-0.5,1.5,c='black',alpha=0.8)
        ax.set_ylim(-2.5,4.5)
        
        font['size']=20
        words = scen.split(' ')
        title = words[0] + ' ' + words[1] + '\n'
        for w in words[2:]:
            title+= w + ' '

        font['size']=14
        if i == 0:
            ax.set_title(title,fontdict=font)
            
        if i == 3:
            ax.set_xticks([0,10,20,30,40,50])
            ax.set_xticklabels(['0','10','20','30','40','50'],fontdict=font)
            ax.set_xlabel('Ensemble Members',fontdict=font)
            
        if j == 0:
            ax.set_yticks([-2,-1,0,1,2,3,4])
            ax.set_yticklabels(['-2','-1','0','1','2','3','4'],fontdict=font)
            ax.set_ylabel(season_titles[season],fontdict=font)
        font['size']=20
        
plt.tight_layout(pad=1)
fig.patch.set_alpha(1)

plt.show()
fig.savefig('/home/queenle/hydro_attribution/dissertation/figures/ECHAM5_tas_nat_trend_test_results.png',dpi=300)
```
