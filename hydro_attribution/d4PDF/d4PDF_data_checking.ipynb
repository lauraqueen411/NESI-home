```python
import numpy as np
import pandas as pd
import xarray as xr
import matplotlib.pyplot as plt
import matplotlib
```

```python
'''
plot raw and gaussian time series for each ensemble member
'''

path = '/nesi/project/niwa00015/queenle/data/hydro_fingerprinting/model/d4PDF/'

for var in ['ROF','TA','PRECIPI']:
    print(var)
    for ens_mem in range(1,101):
        
        print(ens_mem)
        
        fig,axs = plt.subplots(4,2, figsize=(12,10),sharex=True)
        axs = axs.flatten()
        
        if ens_mem == 100:
            number = '100'
        else:
            number = '00' + str(ens_mem) if ens_mem < 10 else '0' + str(ens_mem)

        for i,season in enumerate(['summer','autumn','winter','spring']):    

            all_gaussian = xr.open_dataarray(path + "gaussian/"+var+"/full/" + var + '_Amon_d4PDF_HPB_m' + number + '_gaussian_'+season+'_1951-2021.nc')
            all_raw = xr.open_dataarray(path + "raw/"+var+"/full/" + var + '_Amon_d4PDF_HPB_m' + number + '_raw_'+season+'_1951-2021.nc')

            all_gaussian.mean(['lat','lon']).plot(ax=axs[i*2],c='red',linewidth=4)
            all_raw.mean(['lat','lon','lev']).plot(ax=axs[i*2].twinx(),c='blue',linestyle='--')

            nat_gaussian = xr.open_dataarray(path + "gaussian/"+var+"/full/" + var + '_Amon_d4PDF_HPB_NAT_m' + number + '_gaussian_'+season+'_1951-2021.nc')
            nat_raw = xr.open_dataarray(path + "raw/"+var+"/full/" + var + '_Amon_d4PDF_HPB_NAT_m' + number + '_raw_'+season+'_1951-2021.nc')

            nat_gaussian.mean(['lat','lon']).plot(ax=axs[i*2+1],c='green',linewidth=4)
            nat_raw.mean(['lat','lon','lev']).plot(ax=axs[i*2+1].twinx(),c='cyan',linestyle='--')

            axs[i*2].set_ylabel(season)
            
            all_gaussian.close()
            all_raw.close()
            nat_gaussian.close()
            nat_raw.close()

        plt.savefig('/nesi/project/niwa00015/queenle/plots/d4PDF/ensemble_members/timeseries/'+var+'/full/'+var+'_m'+number+'_timeseries.png')
        plt.close(fig)
```

```python
import scipy.stats
import pymannkendall as mk

def get_linregress_slope(arr):
    if (np.isnan(arr)).any():
        result = np.nan
    else:
        result = np.polyfit([i for i in range(len(arr))],arr,deg=1)[0]
        
    return(result)

def get_variance(arr):
    return(scipy.stats.variation(arr))

def get_mannkendall_slope(arr):
    if (np.isnan(arr)).any():
        result = np.nan
    else:
        result = mk.original_test(arr).slope
        
    return(result)
   
def get_mannkendall_p(arr):
    if (np.isnan(arr)).any():
        result = np.nan
    else:
        result = mk.original_test(arr).p
        
    return(result)
```

```python
'''
Plot trend maps seasonally for each ensemble member
'''

path = '/nesi/project/niwa00015/queenle/data/hydro_fingerprinting/model/d4PDF/'
cmap = matplotlib.cm.get_cmap('viridis_r', 4)

for var in ['ROF','TA','PRECIPI']:#,'ROFS']:
    print(var)
    for ens_mem in range(1,101):
        
        print(ens_mem)
        
        fig,axs = plt.subplots(4,4, figsize=(12,10),sharex=True,sharey=True)
        axs = axs.flatten()
        
        if ens_mem == 100:
            number = '100'
        else:
            number = '00' + str(ens_mem) if ens_mem < 10 else '0' + str(ens_mem)

        for i,season in enumerate(['summer','autumn','winter','spring']):    

            all_da = xr.open_dataarray('/nesi/project/niwa00015/queenle/data/hydro_fingerprinting/model/d4PDF/gaussian/'+var+'/full/'+var+'_Amon_d4PDF_HPB_m'+number+'_gaussian_'+season+'_1951-2021.nc')
            nat_da = xr.open_dataarray('/nesi/project/niwa00015/queenle/data/hydro_fingerprinting/model/d4PDF/gaussian/'+var+'/full/'+var+'_Amon_d4PDF_HPB_NAT_m'+number+'_gaussian_'+season+'_1951-2021.nc')
            ant_da = all_da - nat_da

            xr.apply_ufunc(get_linregress_slope, all_da, input_core_dims=[["time"]], vectorize = True).plot(ax=axs[4*i],vmin=-0.01,vmax=0.01,cmap='RdBu_r')
            xr.apply_ufunc(get_linregress_slope, nat_da, input_core_dims=[["time"]], vectorize = True).plot(ax=axs[4*i+1],vmin=-0.01,vmax=0.01,cmap='RdBu_r')
            xr.apply_ufunc(get_linregress_slope, ant_da, input_core_dims=[["time"]], vectorize = True).plot(ax=axs[4*i+2],vmin=-0.01,vmax=0.01,cmap='RdBu_r')
            im = xr.apply_ufunc(get_mannkendall_p, ant_da, input_core_dims=[["time"]], vectorize = True).plot(ax=axs[4*i+3],vmin=0,vmax=0.4,cmap=cmap)#,norm=norm)

            all_da.close()
            nat_da.close()

        for ax in axs:
            ax.set_xlabel('')
            ax.set_ylabel('')

        axs[0].set_title('All trend')
        axs[1].set_title('Nat trend')
        axs[2].set_title('Ant trend')
        axs[3].set_title('Ant p-value')

        axs[0].set_ylabel('summer')
        axs[4].set_ylabel('autumn')
        axs[8].set_ylabel('winter')
        axs[12].set_ylabel('spring')
        
        plt.savefig('/nesi/project/niwa00015/queenle/plots/d4PDF/ensemble_members/maps/'+var+'/full/'+var+'_m'+number+'_trend_maps.png')
        plt.close(fig)
```
