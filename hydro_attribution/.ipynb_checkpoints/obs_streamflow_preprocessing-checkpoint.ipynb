```python
import numpy as np
import pandas as pd
import xarray as xr
import matplotlib.pyplot as plt
import geopandas as gpd
import rioxarray
#from bias_correction import BiasCorrection, XBiasCorrection
from scipy.stats import percentileofscore
import scipy
```

```python
# load streamflow data
flow_df = pd.read_csv('/nesi/project/niwa00015/queenle/data/flows/seasonal/40_50yrs/40plus.csv')
flow_df['date'] = pd.to_datetime(flow_df['date'])
flow_df = flow_df.set_index('date')
flow_df = flow_df.drop(['4901','1443499'],axis=1)
flow_df = flow_df['1979-03-01':'2019-03-01']

metadata_df = pd.read_csv('/nesi/project/niwa00015/queenle/data/metadata/40_50yr_sites.csv')
metadata_df.set_index('Number',inplace=True)
```

```python
# load catchment shape data
fp = '/nesi/project/niwa00015/queenle/data/arcgis/'
file = 'NESTCAT_v2_NZTM.shp'
catchment_data = gpd.read_file(fp+file)
catchment_data = catchment_data.to_crs("EPSG:4326")
catchment_data = catchment_data.loc[catchment_data['SITE'].isin(flow_df.columns)]
catchment_data.index = catchment_data['SITE']
```

```python
'''
Use quantile mapping to transform observation into gaussian distribution
'''

seasons_a = ['autumn','winter','spring','summer']

for i,s in enumerate([3,6,9,12]):
    
    # select seasonal time series, fill missing data with local mean, then subtract local mean.
    season_df = flow_df.loc[flow_df.index.month == s]
    season_df = season_df.fillna(season_df.mean())
    #season_df = season_df - season_df.mean()

    # create container for corrected values
    corrected_df = season_df.copy()
    
    for site in season_df:

        site_ts = season_df[site].values

        corrected = site_ts.copy()
        for j, val in enumerate(site_ts):
            # find percentile of value within the distribution of observations
            percentile = percentileofscore(site_ts, val, kind= 'mean')
            if percentile == 100 or percentile == 0:
                print
            # replace with corresponding percentile value in norm distribution
            corrected[j] = scipy.stats.norm.ppf(percentile/100, loc=0, scale=1)
            
        corrected_df[site] = corrected
        
    corrected_df.to_csv("/nesi/project/niwa00015/queenle/data/hydro_fingerprinting/obs/obs_"+seasons_a[i]+"_gaussian.csv")
    
```
