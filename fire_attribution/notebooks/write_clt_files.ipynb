```python
import os
import xarray as xr
```

```python
'''
Write all of the monthly control files for individual ensemble members.
'''

for var in ['TA','ROF','PRECIPI']:
    
    f = open('/nesi/nobackup/niwa02986/queenle/d4PDF/'+var+'_mon_sample.ctl')
    lines = f.readlines()
    f.close()
    
    for scen in ['HPB','HPB_NAT']:
        
        for num in range(1,101):
            num_string = '0' + str(num) if num > 9 else '00' + str(num)
            if num == 100:
                num_string = '100'

            new_lines = lines
            
            if scen == 'HPB':
                new_lines[0] = 'DSET ^../monthly/'+var+'_mon_HPB_m'+num_string+'.grd\n'
                new_lines[2] = 'TITLE sfc_avr_mon_HPB_m'+num_string+'    AVR ( Time average data )\n'

                new_f = open('/nesi/nobackup/niwa02986/queenle/d4PDF/HPB/ctl_files/'+var+'_mon_HPB_m'+num_string+'.ctl', 'w')
                for line in new_lines:
                    new_f.write(line)
                new_f.close()
                
            if scen == 'HPB_NAT':
                new_lines[0] = 'DSET ^../monthly/'+var+'_mon_HPB_NAT_m'+num_string+'.grd\n'
                new_lines[2] = 'TITLE sfc_avr_mon_HPB_NAT_m'+num_string+'    AVR ( Time average data )\n'

                new_f = open('/nesi/nobackup/niwa02986/queenle/d4PDF/NAT/ctl_files/'+var+'_mon_HPB_NAT_m'+num_string+'.ctl', 'w')
                for line in new_lines:
                    new_f.write(line)
                new_f.close()
        
```

```python
'''
Write all of the daily control files for individual ensemble members.
'''


for var in ['TA_max','RHA_min','PRECIPI','WIND']:
    
    f = open('/nesi/nobackup/niwa02986/queenle/d4PDF/'+var+'_day_sample.ctl')
    lines = f.readlines()
    f.close()
    
    for scen in ['HPB','HPB_NAT']:
        for num in range(1,101):
            num_string = '0' + str(num) if num > 9 else '00' + str(num)
            if num == 100:
                num_string = '100'

            new_lines = lines
            
            if scen == 'HPB':
                new_lines[0] = 'DSET ^../daily/'+var+'_day_HPB_m'+num_string+'.grd\n'
                new_lines[2] = 'TITLE sfc_avr_day_HPB_m'+num_string+'    AVR ( Time average data )\n'

                new_f = open('/nesi/nobackup/niwa02986/queenle/d4PDF/HPB/ctl_files/'+var+'_day_HPB_m'+num_string+'.ctl', 'w')
                for line in new_lines:
                    new_f.write(line)
                new_f.close()
                
            if scen == 'HPB_NAT':
                new_lines[0] = 'DSET ^../daily/'+var+'_day_HPB_NAT_m'+num_string+'.grd\n'
                new_lines[2] = 'TITLE sfc_avr_day_HPB_NAT_m'+num_string+'    AVR ( Time average data )\n'

                new_f = open('/nesi/nobackup/niwa02986/queenle/d4PDF/NAT/ctl_files/'+var+'_day_HPB_NAT_m'+num_string+'.ctl', 'w')
                for line in new_lines:
                    new_f.write(line)
                new_f.close()
                

        
```

```python
'''
Convert all monthly files from binary to netcdf, .nc files saved in /nesi/project/niwa02986/queenle/data/d4pdf/ directory.
'''

for var in ['TA','ROF','PRECIPI']:
    for scen in ['HPB','HPB_NAT']:
        for num in range(1,101):
            num_string = '0' + str(num) if num > 9 else '00' + str(num)
            if num == 100:
                num_string = '100'

            if scen == 'HPB':
                in_file = '/nesi/nobackup/niwa02986/queenle/d4PDF/HPB/ctl_files/'+var+'_mon_HPB_m'+num_string+'.ctl'
                out_file = '/nesi/project/niwa02986/queenle/data/d4pdf/global/HPB/mon/' + var +'/'+var+'_Amon_d4PDF_HPB_m'+num_string+'_2011-2021.nc'

                os.system("cdo -f nc import_binary " + in_file + " " + out_file)

            if scen == 'HPB_NAT':
                in_file = '/nesi/nobackup/niwa02986/queenle/d4PDF/NAT/ctl_files/'+var+'_mon_HPB_NAT_m'+num_string+'.ctl'
                out_file = '/nesi/project/niwa02986/queenle/data/d4pdf/global/HPB_NAT/mon/' + var +'/'+var+'_Amon_d4PDF_HPB_NAT_m'+num_string+'_2011-2021.nc'

                os.system("cdo -f nc import_binary " + in_file + " " + out_file)

```

```python
'''
Convert all daily files from binary to netcdf, .nc files saved in /nesi/project/niwa02986/queenle/data/d4pdf/ directory.
'''

var = 'TA_max' #,'RHA_min','PRECIPI','WIND']:

for scen in ['HPB','HPB_NAT']:
    for num in range(1,101):
        num_string = '0' + str(num) if num > 9 else '00' + str(num)
        if num == 100:
            num_string = '100'

        if scen == 'HPB':
            in_file = '/nesi/nobackup/niwa02986/queenle/d4PDF/HPB/ctl_files/'+var+'_day_HPB_m'+num_string+'.ctl'
            out_file = '/nesi/project/niwa02986/queenle/data/d4pdf/global/HPB/day/' + var +'/'+var+'_Aday_d4PDF_HPB_m'+num_string+'_2011-2021.nc'

            os.system("cdo -f nc import_binary " + in_file + " " + out_file)

        if scen == 'HPB_NAT':
            in_file = '/nesi/nobackup/niwa02986/queenle/d4PDF/NAT/ctl_files/'+var+'_day_HPB_NAT_m'+num_string+'.ctl'
            out_file = '/nesi/project/niwa02986/queenle/data/d4pdf/global/HPB_NAT/day/' + var +'/'+var+'_Aday_d4PDF_HPB_NAT_m'+num_string+'_2011-2021.nc'

            os.system("cdo -f nc import_binary " + in_file + " " + out_file)

```
