```python
import xarray as xr
from datetime import datetime
import math
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from scipy import stats
```

```python
hr_pr_FWI = xr.open_dataset("/nesi/project/niwa00015/queenle/data/fire/d4pdf/testing_hr_v_day_pr/d4pdf_FWI_hrly_pr.nc")
day_pr_FWI = xr.open_dataset("/nesi/project/niwa00015/queenle/data/fire/d4pdf/testing_hr_v_day_pr/d4pdf_FWI_daily_pr.nc")
```

```python
hr_pr_FWI.FWI.mean('time').plot()
```

```python
mp = day_pr_FWI.FWI.mean('time')
```

```python
import scipy 

fig,ax = plt.subplots(3,2,figsize=(8,12))
axs = ax.flatten()

###
lon,lat = (hr_pr_FWI.FWI.sel(latitude=60,longitude=-120,method='nearest').longitude.values.tolist(),hr_pr_FWI.FWI.sel(latitude=60,longitude=-120,method='nearest').latitude.values.tolist())
hr_ts = hr_pr_FWI.FWI.sel(latitude=60,longitude=-120,method='nearest').sel(time=slice('1950','2011')).values.flatten()
day_ts = day_pr_FWI.FWI.sel(latitude=60,longitude=-120,method='nearest').values.flatten()

axs[0].scatter(hr_ts,day_ts)
x = [i for i in range(50)]
axs[0].plot(x, x, c='black', label='slope=1')
reg = scipy.stats.linregress(hr_ts,day_ts)
axs[0].plot(x, [m*reg.slope + reg.intercept for m in x], c='red', label='slope=' +str(reg.slope))
axs[0].set_xlabel('hourly')
axs[0].set_ylabel('daily')
axs[0].legend()
mp.plot(ax=axs[1])
axs[1].text(lon,lat,'x',color='red',size=14)
axs[1].set_title('')
axs[0].text(0.05,0.7,'d',transform=axs[0].transAxes,size=18,color='green')

###
lon,lat = (hr_pr_FWI.FWI.sel(latitude=40,longitude=-120,method='nearest').longitude.values.tolist(),hr_pr_FWI.FWI.sel(latitude=40,longitude=-120,method='nearest').latitude.values.tolist())
hr_ts = hr_pr_FWI.FWI.sel(latitude=40,longitude=-120,method='nearest').sel(time=slice('1950','2011')).values.flatten()
day_ts = day_pr_FWI.FWI.sel(latitude=40,longitude=-120,method='nearest').values.flatten()

axs[2].scatter(hr_ts,day_ts)
x = [i for i in range(125)]
axs[2].plot(x, x, c='black', label='slope=1')
reg = scipy.stats.linregress(hr_ts,day_ts)
axs[2].plot(x, [m*reg.slope + reg.intercept for m in x], c='red', label='slope=' +str(reg.slope))
axs[2].set_xlabel('hourly')
axs[2].set_ylabel('daily')
axs[2].legend()
mp.plot(ax=axs[3])
axs[3].text(lon,lat,'x',color='red',size=14)
axs[3].set_title('')
axs[2].text(0.05,0.7,'e',transform=axs[2].transAxes,size=18,color='green')

###
lon,lat = (hr_pr_FWI.FWI.sel(latitude=33,longitude=-115,method='nearest').longitude.values.tolist(),hr_pr_FWI.FWI.sel(latitude=33,longitude=-115,method='nearest').latitude.values.tolist())
hr_ts = hr_pr_FWI.FWI.sel(latitude=33,longitude=-115,method='nearest').sel(time=slice('1950','2011')).values.flatten()
day_ts = day_pr_FWI.FWI.sel(latitude=33,longitude=-115,method='nearest').values.flatten()

axs[4].scatter(hr_ts,day_ts)
x = [i for i in range(145)]
axs[4].plot(x, x, c='black', label='slope=1')
reg = scipy.stats.linregress(hr_ts,day_ts)
axs[4].plot(x, [m*reg.slope + reg.intercept for m in x], c='red', label='slope=' +str(reg.slope))
axs[4].set_xlabel('hourly')
axs[4].set_ylabel('daily')
axs[4].legend()
mp.plot(ax=axs[5])
axs[5].text(lon,lat,'x',color='red',size=14)
axs[5].set_title('')
axs[4].text(0.05,0.7,'f',transform=axs[4].transAxes,size=18,color='green')

plt.savefig('examples_2.png',dpi=300)
```
